# Import all the necessary libraries and Data Forlder

```python
from pathlib import Path
from plot.plot import ExperimentPlot
from IPython.display import Markdown, display
```

```python
project_base = Path().resolve()
data_dir = project_base.joinpath('data')
memory_change_dir = data_dir.joinpath('003-MemoryChange')
simulator_assessment_dir = data_dir.joinpath('004-SimulatorAssessment')
anova_dir = data_dir.joinpath('001-Anova')
monte_carlo_dir = data_dir.joinpath('002-Monte Carlo')
```

```python
# Please change the output directory to your desired location
# if you are not happy with exporting images to the current directory
# which is at <project base>/evaluation/output
output_dir = project_base.joinpath('output')
output_dir.mkdir(exist_ok=True, parents=True)
view_mode = False
```

```python
e_plot = ExperimentPlot(output_dir=output_dir)
```

# Summary of All Experimental Results

```python
df_a, anova_summary = e_plot.summarize_data(data_dir)
display(Markdown(df_a.to_markdown(index=False)))
```

```python
display(Markdown(anova_summary.to_markdown(index=False)))
```

# Figrues for Introduction
Change in memory between two iterations is influenced by where the zeroing gradient function, optimizer.zero grad, is placed. Position 0 (POS0) means calling the function at the beginning of each iteration, while Position 1 (POS1) means calling it just before backward propagation

```python
fig = e_plot.plot_memory_fluctuation_by_changing_zero_out(
    "Motivation-Memory Change Impacted by zero out operation", 
    memory_change_dir,
    image_size=[1500, 500],
    font_size=30,
    legend_font_size=30,
    tickfont_size=30,
    line_width=4,
    view_mode=view_mode
)
fig.show()
```

# CUDACachingAllocator Simulator Assessment Result 

```python
fig = e_plot.plot_assessment_result_between_snapshot_and_simulation(
    title="Design-Comparison of Memory Allocation between Snapshot and Simulation",
    data_dir=simulator_assessment_dir,
    image_size=(1500, 500),
    font_size=35,
    legend_font_size=30,
    tickfont_size=25,
    view_mode=view_mode
)
fig.show()
```

# ANOVA Experimental Results

## Statistic for ANOVA Data

```python
df_a, anova_summary = e_plot.summarize_data(anova_dir)
display(Markdown(df_a.to_markdown(index=False)))
```

## Box Plot of Relative Error for CNN Models

```python
filename = "Evaluation-MRE in box plot"
overall_median = True
image_size = (1000, 400)
font_size = 28
lengend_font_size = 24
tickfont_size = 24
fig = e_plot.plot_relative_error_in_box_diagram_with_verification_data(
    title=f"{filename}-CNN",
    data_dir=anova_dir.joinpath("CNN"),
    overall_median=overall_median,
    image_size=image_size,
    font_size=font_size,
    legend_font_size=lengend_font_size,
    tickfont_size=tickfont_size,
    yrange=(0, 150),
    view_mode = view_mode,
	median_box_offset=(0.62, 0.95, 0.11),
    legend_position=(0.83, 0.58)
)
fig.show()

fig = e_plot.plot_relative_error_in_box_diagram_with_verification_data(
    title=f"{filename}-Transformer",
    data_dir=anova_dir.joinpath("Transformer"),
    overall_median=overall_median,
    image_size=image_size,
    font_size=font_size,
    legend_font_size=lengend_font_size,
    tickfont_size=tickfont_size,
    yrange=(0, 150),
    view_mode = view_mode,
    median_box_offset=(0.93, 0.72, 0.14),
	legend_position=(1.2, 0.1)
)
fig.show()

```

# Four-Quadrant Analysis
- Bottom left represents an optimal quadrant
- Bottom right: represents an underestimation quadrant
- Top left: represents an overestimation quadrant
- Top right: represents an worst quadrant

```python
fig = e_plot.plot_probability_estimation_vs_error_scatter_diagram_model_base(
    title="Evaluation-MRE vs PEF-ANOVA",
    data_dir=anova_dir,
    image_size=(800, 800),
    title_font_size=35,
    legend_font_size=30,
    tickfont_size=25,
    quadrant_font_size=40,
    marker_size=16,
    view_mode=view_mode
)
fig.show()
```
