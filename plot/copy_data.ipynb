```python
import shutil
import tqdm
from pathlib import Path
from ures.files import filter_files
```

```python
data_dir = Path("/Volumes/ResearchData/101-xMem-LLM/005-LLM Experiments/005-new experiments/005-CoLab")
data_dir
```

```python
result_files = filter_files("evaluation_result.json", data_dir, fuzz=False)
len(result_files)
```

```python
current_dir = Path().resolve()
target_data_dir = current_dir.joinpath("data_new")
target_data_dir.mkdir(parents=True, exist_ok=True)
```

```python
for src_file in tqdm.tqdm(result_files):
    partial_path = src_file.split(str(data_dir))[-1][1:]
    new_target_file = target_data_dir.joinpath(partial_path)
    new_target_file.parent.mkdir(parents=True, exist_ok=True)
    shutil.copy2(src_file, new_target_file)
```
