```python
import shutil
all_dirs = os.listdir(project_base)
max_batch = config.batch_range[1]
print(f"All batch over {max_batch} will be removed")
for name in all_dirs:
	split_name = name.split('_')
	if int(split_name[-3]) >= max_batch:
		print(name)
		shutil.rmtree(project_base.joinpath(name))
```
